### Bayesian Regression

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import pymc
```

```python
# Create some random data to fit 
x = np.linspace(0, 10, 500)
y = 4 + 2 * x + np.random.normal(loc=0, scale=1, size=len(x)) * 2


# Simple linear regression
reg = np.polyfit(x, y, deg=1)
inter = reg[1]
slope = reg[0]

fig, ax = plt.subplots(1, 1, figsize=(8,5))
ax.scatter(x, y, c=y, marker='v', cmap='coolwarm')
ax.plot(x, inter + slope * x, color='black')
ax.set_xlabel('x')
ax.set_ylabel('y')
plt.show()
```

We obtain as results the slope and the intecept of the regression line via OLS
