```python
# Convert Date into different format
df['Date'] = pd.to_datetime(df['Date'], format='%d-%b-%y')
# Sort data according to Date and Ticker
df = df.sort_values(by=['Ticker', 'Date'], ascending=True).reset_index(drop=True)

# Use Date as new Index
df = df.set_index('Date')
df
```

```python
# Daily percentage return
# df['Return'] = df.groupby('Ticker')['Close'].transform(lambda x: (x - x.shift(1)) / x.shift(1))
df['PctReturn'] = df.groupby('Ticker')['Close'].pct_change()

# Daily log-return
df['LogReturn'] = df.groupby('Ticker')['Close'].transform(lambda x: np.log(x / x.shift(1)))

# Cumulative return (1 + r_1) * (1 + r_2) * ...
df['CumReturn'] = df.groupby('Ticker')['PctReturn'].transform(lambda x: (1 + x).cumprod() - 1)

df
```

```python
fig, ax = plt.subplots(1, 3, figsize=(18, 4))
df.groupby('Ticker')['PctReturn'].plot(ax=ax[0], ylabel='PctReturn', legend=True)
df.groupby('Ticker')['LogReturn'].plot(ax=ax[1], ylabel='LogReturn', legend=True)
df.groupby('Ticker')['CumReturn'].plot(ax=ax[2], ylabel='CumReturn', legend=True)
plt.tight_layout()
plt.show()
```

#### WINDOW FUNCTIONS

```python
# Moving average
df['MA_Close_5d'] = df.groupby('Ticker')['Close'].transform(lambda x: x.rolling(window=5).mean())
df['MA_Close_20d'] = df.groupby('Ticker')['Close'].transform(lambda x: x.rolling(window=20).mean())

df['MA_PctReturn_5d'] = df.groupby('Ticker')['PctReturn'].transform(lambda x: x.rolling(window=5).mean())
df['MA_PctReturn_20d'] = df.groupby('Ticker')['PctReturn'].transform(lambda x: x.rolling(window=20).mean())

# Rolling volatility
df['Vol_5d'] = df.groupby('Ticker')['Close'].transform(lambda x: x.rolling(window=5).std())
df['Vol_20d'] = df.groupby('Ticker')['Close'].transform(lambda x: x.rolling(window=20).std())

# Rolling Sharpe Ratio
tradingDays = 252
rf = 0.05 / tradingDays
df['Sharpe_5d'] = (df['MA_PctReturn_5d'] - rf) / df['Vol_5d']
df['Sharpe_20d'] = (df['MA_PctReturn_20d'] - rf) / df['Vol_20d']

df
```

```python
fig, ax = plt.subplots(1, 2, figsize=(12, 4))
df.groupby('Ticker')['MA_PctReturn_5d'].plot(ax=ax[0], ylabel='MA(PctReturn, 5 days)', legend=True)
df.groupby('Ticker')['MA_PctReturn_20d'].plot(ax=ax[1], ylabel='MA(PctReturn, 20 days)', legend=True)
plt.tight_layout()
plt.show()
```

#### RESAMPLING

Resampling requires rules to specify, some common rules include:
- "`D`"     Daily
- "`B`"	    Business day
- "`W`"	    Weekly
- "`W-FRI`"	Weeks ending Friday
- "`ME`"	Month end
- "`MS`"	Month start
- "`QE`"	Quarter end
- "`QS`"	Quarter start
- "`YE`"	Year end
- "`YS`"	Year start
- "`H`"	    Hourly
- "`15min`"	Every 15 minutes
- "`30s`"	Every 30 seconds

```python
# We have daily returns --> weekly
weekly = df.groupby('Ticker').resample('W-FRI').agg({'Close': 'last', 'Volume': 'sum'})
weekly
```

```python
# We have daily returns --> monthly
monthly = df.groupby('Ticker').resample('ME').agg({'Close': 'last', 'Volume': 'sum'})
monthly
```
