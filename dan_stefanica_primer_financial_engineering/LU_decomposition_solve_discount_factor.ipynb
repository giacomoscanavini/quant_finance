### LU DECOMPOSITION

We can use LU decomposition to find the discount factors knowing cash flows form several bonds
1. 11 months semiannual with coupon rate 4% and bond price $101.5
2. 17 months annual with coupon rate 5% and bond price $105.5
3. 23 months semiannual with coupon rate 3% and bond price $101
4. 23 months annual with coupon rate 6% and bond price $106.75

First we need to identify the cash flow a build a matrix whose columns correspond to each cash flow periods

Note: semiannual coupon rates apply for half value of the coupon

Assuming at maturity we get F+C for each bond and C at each annual or semiannual period before maturity (5   11   17   23 months):
1. [C1,  F+C1, 0,    0   ] with C1 = 2
2. [C2,  0,    F+C2, 0   ] with C2 = 5
3. [C3,  C3,   C3,   F+C3] with C3 = 1.5
4. [0,   C4,   0,    F+C4] with C4 = 6

```python
import numpy as np
import scipy 

CF = np.array([[2,   102, 0,   0    ],
               [5,   0,   105, 0    ],
               [1.5, 1.5, 1.5, 101.5],
               [0,   6,   0,   106  ]])

y = np.array([101.5, 105.5, 101, 106.75])

periods = np.array([5/12, 11/12, 17/12, 23/12])
```

```python
# In LU decomposition we are considering (Permulation matrix P, Lower triangular matrix L, Upper triangular matrix U)
# The initial equation is given by Ax = y
# LUx = Py
# Lb = Py --> b = L^-1 P y
# Ux = b  --> x = U^-1 b = U^-1 L^-1 P y

P, L, U = scipy.linalg.lu(CF)

x = np.linalg.inv(U) @ np.linalg.inv(L) @ P @ y
```

The solution to the equation $x$ corresponds to the discount rates $e^{-rt}$ at different times $t = {\frac{5}{12}, \frac{11}{12}, \frac{17}{12}, \frac{23}{12}}$

```python
rates = np.array([round(-np.log(val)/t * 100, 2) for val,t in zip(x, periods)])

for r,t in zip(rates, periods):
    print(f"r(0, {t*12} months) = {r}%")
```
