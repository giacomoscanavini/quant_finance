### Numerical Methods for Integration

Numerical integration can be achieved using the following methods:
1) Midpoint (Rectangles with height given by segment mid-point - error O($n^2$))
2) Trapezoid (Similar to midpoint rule but use linear segment to connect interval endpoints - error O($n^2$))
3) Simpson's Rule (More sophisticated method with lowest error term among 3 methods - error O($n^4$))

#### 1D Case
Integrate $x^2$ between 0 and 1, the solution is 1/3

```python
import numpy as np
import scipy


def f(x):
    return x**2


a = 0
b = 1
n = 10

x = np.linspace(a, b, n + 1)
y = f(x)

# Midpoint rule (scipy does not have a direct midpoint-rule function)
h = (b - a) / n
midpoints = a + (np.arange(n) + 0.5) * h
mid = h * np.sum(f(midpoints))

# Trapezoidal rule
trap = np.trapezoid(y, x)
trap = scipy.integrate.trapezoid(y, x)

# Simpson's rule
simp = scipy.integrate.simpson(y, x=x)

# High-accuracy adaptive quadrature (Additional method)
quad_value, quad_error = scipy.integrate.quad(f, a, b)

exact = 1 / 3

print("Exact:      ", exact)
print("Midpoint:   ", mid, "error:", abs(mid - exact))
print("Trapezoid:  ", trap, "error:", abs(trap - exact))
print("Simpson:    ", simp, "error:", abs(simp - exact))
print("Quad:       ", quad_value, "estimated error:", quad_error)
```

#### Double integration
This time integrate $f(x, y) = x^2 + y^2$ over $x \in [0, 1]$ and $y \in [0, 2]$

```python
def f(y, x):
    return x**2 + y**2

x_min, x_max = 0, 1
y_min, y_max = 0, 2

result, error = scipy.integrate.dblquad(
    f,
    x_min,
    x_max,
    y_min,
    y_max,
)

print("2D integral:", result)
print("Estimated error:", error)
```

```python
# Note that this time we change the integration inteval for y in [0, 1-2x]

def f(y, x):
    return x**2 + y**2

x_min, x_max = 0, 1
y_min, y_max = 0, 2

result, error = scipy.integrate.dblquad(
    f,
    x_min,
    x_max,
    lambda x: y_min,
    lambda x: 1 - 2*x,
)

print("2D integral:", result)
print("Estimated error:", error)
```

#### Triple integration
Now consider a triple integration over x, y, and z

```python
def f(x, y, z):
    return x**2 + y**2 + z**2

x_min, x_max = 0, 1
y_min, y_max = 0, 2
z_min, z_max = 0, 4

result, error = scipy.integrate.tplquad(
    f,
    x_min, x_max,
    y_min, y_max,
    z_min, z_max,
)

print("3D integral:", result)
print("Estimated error:", error)
```

```python
f = lambda z, y, x: x*y*z
scipy.integrate.tplquad(
    f, 
    0, 1, 
    0, lambda x: 1-2*x, 
    0, lambda x, y: 1-x-2*y,
)
```
