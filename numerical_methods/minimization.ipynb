### Numerical Minimization / Maximization

In general we might want to find the min or max of a function numerically

If we function is differentiable we can take the derivative and set it to 0, but numerically we can use the `scipy.optimize.minimize` method

This means that if we want to find the max we can simply define the function to minimize and then place a minus sign in front

```python
# Simple 1D case with no constraints

import numpy as np
from scipy.optimize import minimize

def f(x):
    return (x[0] - 3)**2 + 2

result = minimize(
    fun=f,
    x0=np.array([0.0])
)

print(f"Minimum location: {result.x}")
print(f"Function at the min: {result.fun}")
```

```python
# Simple 1D case with a constraint given by a simple bound for the solution
# Constraint: x > 5

def f(x):
    return (x[0] - 3)**2 + 2

bounds = [(5, None)]

result = minimize(
    fun=f,
    x0=np.array([6.0]),
    bounds=bounds
)

print(f"Minimum location: {result.x}")
print(f"Function at the min: {result.fun}")
```

```python
# Simple 1D case with a complex constraint, this might be seen as Lagrange multipliers numerical method
# Constraint: x + y - 1 = 0

def f(v):
    x, y = v
    return x**2 + y**2

constraint = {
    "type": "eq",
    "fun": lambda v: v[0] + v[1] - 1
}

result = minimize(
    fun=f,
    x0=np.array([0.0, 0.0]),
    constraints=[constraint]
)

print(f"Minimum location: {result.x}")
print(f"Function at the min: {result.fun}")
```
