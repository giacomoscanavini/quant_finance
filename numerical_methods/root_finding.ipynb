### Root Finding

Numerically, one can find the solution to $f(x) = 0$ using the following methods:
1) Newton (requires $f(x)$ to be differentiable)
2) Secant (alternative to Newton's)
3) Bisection (the simplest of the 3 methods)

In Python, one can rely on the `scipy` library

#### 1D Case
Solve the equation 

$$
f(x) = x^2 - 2 = 0
$$

The solution should be $x = \sqrt{2}$

```python
import numpy as np
from scipy.optimize import root_scalar


# 1D function
def f(x): return x**2 - 2
def df(x): return 2 * x

# Bisection
sol_bisect = root_scalar(
    f,
    bracket=[1, 5],
    method="bisect",
    xtol=1e-8
)
print("Bisection")
print(sol_bisect.root)
print(sol_bisect.iterations)
print(sol_bisect.converged)

# Newton
sol_newton = root_scalar(
    f,
    x0=1,
    fprime=df,
    method="newton",
    xtol=1e-8
)
print("\nNewton")
print(sol_newton.root)
print(sol_newton.iterations)
print(sol_newton.converged)

# Secant
sol_secant = root_scalar(
    f,
    x0=1,
    x1=2,
    method="secant",
    xtol=1e-8
)
print("\nSecant")
print(sol_secant.root)
print(sol_secant.iterations)
print(sol_secant.converged)
```

#### 2D case
Solve a system of equations: 

$$
\begin{align}
x^2 + y^2 -4 &= 0 \\
x - y &= 0
\end{align}
$$

The first equation is a circle of radius 2
The second equation is the line x = y

The solution should be near: $x = y = \sqrt{2}$

```python
# 2D function
from scipy.optimize import root

def F(z):
    x, y = z
    f1 = x**2 + y**2 - 4
    f2 = x - y
    return np.array([f1, f2])


x0 = np.array([1.0, 1.0])

sol = root(F, x0)

print(sol.x)
print(sol.fun)
print(sol.success)
print(sol.message)
```
