### Pricing European Options
In a binomial tree model with up and down movements, we have the Cox-Ross-Rubinstein formula that can be used to calculate the price at t=0 for both Call and Put options

```python
import scipy, math
import numpy as np

class PricingOption:
    def __init__(self, S0, d, rf, u, X, nSteps):
        """
        S0: underlying asset value at t=0
        d : stock down movement
        rf: risk-free interest rate
        u : stock up movement
        X : strike price
        nSteps: number of steps in binomial tree model
        """
        self.S0 = S0
        self.d = d
        self.rf = rf
        self.u = u
        self.X = X
        self.nSteps = nSteps

        self.p_star = self.calculate_risk_neutral_prob()
        self.q = self.calculate_q()
        self.m = self.calculate_m()

    def calculate_risk_neutral_prob(self):
        return (self.rf - self.d) / (self.u - self.d)

    def calculate_q(self):
        return self.calculate_risk_neutral_prob() * (1 + self.u) / (1 + self.rf)
    
    def calculate_m(self):
        num = np.log(self.X / self.S0 / (1 + self.d)**self.nSteps)
        den = np.log((1 + self.u) / (1 + self.d))
        return int(math.floor(num / den) + 1)

    def calculate_delta(self):
        print(f"\tDelta = {scipy.stats.binom.sf(k=self.m - 1, n=self.nSteps, p=self.q)}")

    def cost_europe_call(self):
        term1 = self.S0 * scipy.stats.binom.sf(k=self.m - 1, n=self.nSteps, p=self.q)
        term2 = self.X / (1 + self.rf)**self.nSteps * scipy.stats.binom.sf(k=self.m - 1, n=self.nSteps, p=self.p_star)
        return term1 - term2

    def price_call(self):
        print(f"---- European Call Option Pricing ----")
        print(f"\tp* = {self.p_star}")
        print(f"\tq = {self.q}")
        print(f"\tm = {self.m}")
        print(f"\tCall Price = {self.cost_europe_call()}")

    def cost_europe_put(self):
        term1 = self.S0 * scipy.stats.binom.cdf(k=self.m - 1, n=self.nSteps, p=self.q)
        term2 = self.X / (1 + self.rf)**self.nSteps * scipy.stats.binom.cdf(k=self.m - 1, n=self.nSteps, p=self.p_star)
        return term2 - term1

    def price_put(self):
        print(f"---- European Put Option Pricing ----")
        print(f"\tp* = {self.p_star}")
        print(f"\tq = {self.q}")
        print(f"\tm = {self.m}")
        print(f"\tPut Price = {self.cost_europe_put()}")
```

```python
S0, d, rf, u, X, nSteps = 50, -0.1, 0.05, 0.3, 60, 3
option = PricingOption(S0, d, rf, u, X, nSteps)
option.price_call()
option.calculate_delta()
option.price_put()
```
