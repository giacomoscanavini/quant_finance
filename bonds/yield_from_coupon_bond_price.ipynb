### Yield of a Coupon Bond
The yield of a bond is defined as the rate such that 

$$
P_0 = \sum_{t=1}^{T} \frac{C_t}{(1 + r_{0,t})^t} = \sum_{t=1}^{T} \frac{C_t}{(1 + Y)^t} 
$$


To find the yield of a coupon bond we need to consider the function 

$$
f(Y) = \sum_{t=1}^{T} \frac{C_t}{(1 + Y)^t} - P_0
$$

and use a method such as Newton's or Bisect to find its root to solve for `Y`

```python
import scipy
import numpy as np

def yield_from_coupon_price(y, bond_value=75, face_value=100, years_to_maturity=3, coupon_rate=0.05, coupon_frequency=1):
    nPeriods = int(years_to_maturity * coupon_frequency)
    couponValue = face_value * coupon_rate / coupon_frequency
    y = y / coupon_frequency

    cash_flows = np.full(nPeriods, couponValue)
    cash_flows[-1] += face_value
    times = np.arange(1, nPeriods + 1)
    price = np.sum(cash_flows / (1 + y)**times)
    return price - bond_value

results = scipy.optimize.bisect(
    f=yield_from_coupon_price, 
    a=0,
    b=1,
    args=(75, 100, 3, 0.05, 1),
)

print(f"The yield is {round(results * 100, 2)}%")
```
